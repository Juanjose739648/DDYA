### 5) Análisis de complejidad **[15 pts]**
Incluye:
- Justificación
- Función T[n]
- Mejor y peor caso con gráfica


- Justificacion:
La recuersion simple genera calculos redundantes con un tiempo exponencial 0(2^n). Al aplicar memoizacion guardamos los resultados previos en una cache para que cada mes se calcule una sola vez. Asi, las consultas futuras se resuelven al instante en tiempo constante
- Funcion T[n]
Es lineal:
T(n) = O(n)
- Mejor y peor caso 

Mejor: O(1), ocurre si consultamos un caso base o un mes que ya esta guardado en la cache 

Peor: O(n), ocurre la primera vez que consultamos un mes nuevo, ya que debe calcular los meses uno por uno 

## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
def produccion_vivero(n, memo=None):
   if memo is None:
       memo = {}
   if n == 0:
       return 0
   if n == 1:
       return 1
   if n in memo:
       return memo[n]
   memo[n] = produccion_vivero(n - 1, memo) + produccion_vivero(n - 2, memo)
   return memo[n]

def main():
   entrada = input("Ingrese el mes a consultar: ")
   if "." in entrada:
       print("Error: Solo se permiten números enteros.")
       return
   try:
       mes = int(entrada)
       if mes == 0:
           print("El cálculo debe ser desde el número 1.")
           return
       if mes < 0:
           print("El mes no puede ser negativo.")
           return
       resultado = produccion_vivero(mes)
       print(f"Plantas producidas en el mes {mes}: {resultado}")
   except ValueError:
       print("Error: Debe ingresar un número entero válido.")


main()
```
